## Busqueda valor máximo

```python
import random
import time

lista = [random.randint(1, 5000000) for r in range(10**6)]
inicio = time.time()
mayor = -1

for n in lista:
    if n > mayor:
        mayor = n


fin = time.time()
print("el valor maxino es:", mayor)
print("el tiempo de ejecucion fue: ",fin - inicio)
```

## Identificar K-dominante

```python
import time
import random
import string

texto = ''.join(random.choices(string.ascii_letters, k=1000))


inicio = time.time()

mejor = ""
puntaje = -1

for option in texto:
  actual = 0;

  for letra in texto:
    if letra == option:
      actual += 1

  if actual > puntaje:
    mejor = option
    puntaje = actual

fin = time.time()




print("La letra mas repetida fue:", mejor)
print("Cantidad: ", puntaje)
print("el tiempo de ejecucion fue: ",fin - inicio)
```
